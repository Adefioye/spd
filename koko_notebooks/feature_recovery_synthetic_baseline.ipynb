# Feature Recovery: Synthetic Baseline

This notebook runs the stage-0 SAE-style calibration experiments from `research_notes/feature_recovery.md` using the local feature-recovery scripts.

```python
from pathlib import Path
import json
import subprocess

REPO_ROOT = Path.cwd()
CONFIG = REPO_ROOT / 'koko_experiments/feature_recovery/configs/sae_baseline_orthogonal.yaml'
SCRIPT = REPO_ROOT / 'koko_experiments/feature_recovery/train_sae_baseline.py'
```

## Run the SAE baseline

```python
cmd = ['python', str(SCRIPT), str(CONFIG)]
result = subprocess.run(cmd, cwd=REPO_ROOT, capture_output=True, text=True, check=True)
print(result.stdout)
out_dir = Path(result.stdout.strip().splitlines()[-1])
out_dir
```

## Inspect saved metrics

```python
metrics = json.loads((out_dir / 'metrics.json').read_text())
metrics
```
