# Feature Recovery: Results Dashboard

Aggregate multiple analyzed SPD runs and visualize direction-recovery trends.

```python
from pathlib import Path
import json
from collections import defaultdict
import matplotlib.pyplot as plt

REPO_ROOT = Path.cwd()
ANALYSIS_ROOT = REPO_ROOT / 'spd_out/spd'
paths = sorted(ANALYSIS_ROOT.rglob('feature_recovery_analysis.json'))
rows = [json.loads(path.read_text()) for path in paths]
len(rows)
```

## Build a compact table

```python
table = []
for row in rows:
    for layer in row['layers']:
        table.append({
            'run': Path(row['spd_run_dir']).name,
            'layer': layer['layer_name'],
            'role': layer['direction_role'],
            'mmc': layer['mean_matched_cosine'],
            'cov95': layer['coverage_at_095'],
            'frag': layer['fragmentation'],
            'merge': layer['merging'],
        })
table[:10]
```

## Plot matched cosine by layer/role

```python
grouped = defaultdict(list)
for row in table:
    grouped[f"{row['layer']}::{row['role']}"].append(row['mmc'])

fig, ax = plt.subplots(figsize=(10, 4))
labels = list(grouped)
values = [sum(v) / len(v) for v in grouped.values()]
ax.bar(range(len(labels)), values)
ax.set_xticks(range(len(labels)))
ax.set_xticklabels(labels, rotation=45, ha='right')
ax.set_ylabel('Mean matched cosine')
ax.set_title('Feature-recovery direction alignment across analyzed runs')
plt.tight_layout()
```
